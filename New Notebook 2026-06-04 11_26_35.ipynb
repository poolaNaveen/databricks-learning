```python
customer_df = spark.read.csv("/Volumes/workspace/default/raw_files/Databricks_files/Databricks_files/customers.csv",
                             header = True, inferSchema= True)
```

```python
order_df = spark.read.csv("/Volumes/workspace/default/raw_files/Databricks_files/Databricks_files/orders.csv",
                          header = True, inferSchema = True)
```

```python
customer_order_df = customer_df.join(order_df,["customer_id"],"inner")
```

```python
customer_order_df.printSchema()
```

```python
customer_order_df.show()
```

```python
from pyspark.sql.functions import col, sum
clean_df = customer_order_df.groupBy("city").agg(sum("unit_price").alias("total_price"))
```

```python
trasform_df = clean_df.orderBy(col("total_price").desc())
```

```python
trasform_df.write.mode("overwrite").csv("/Volumes/workspace/default/raw_files/Databricks_files/Databricks_files/Output_data/business_data")
```
